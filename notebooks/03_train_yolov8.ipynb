```python
from ultralytics import YOLO
import torch
print(torch.__version__)
print('CUDA:',torch.cuda.is_available())
```

## Dataset Validation

```python
DATASET_PATH='/kaggle/input/fingerprint-small-dataset'
```

```python
model=YOLO('yolov8n.pt')
results=model.train(data=f'{DATASET_PATH}/data.yaml',epochs=75,imgsz=640,batch=16,project='runs/detect',name='small_fingerprints_v1')
```

```python
metrics=model.val()
print(metrics)
```
